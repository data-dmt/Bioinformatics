##### Apartado 2a

La replicación del ADN es semiconservativa: cada molécula hija conserva una hebra original y contiene una hebra nueva sintetizada de forma complementaria.

Tras una ronda de replicación se obtienen dos moléculas:

Molécula hija 1:
Original:  5' – ATG CCG TTA GCT – 3'
Nueva:     3' – TAC GGC AAT CGA – 5'

Molécula hija 2:
Nueva:     5' – ATG CCG TTA GCT – 3'
Original:  3' – TAC GGC AAT CGA – 5'

Las nuevas hebras coinciden en secuencia con las hebras complementarias originales porque se aplican las reglas de complementariedad:

A ↔ T
C ↔ G

##### Apartado 2b

Las enzimas implicadas son las siguientes:
- Helicasa: rompe los puentes de hidrógeno entre las bases y separa las dos hebras del ADN.
- Primasa: sintetiza pequeños cebadores de ARN que proporcionan un extremo desde el que puede empezar a trabajar la ADN polimerasa.
- ADN polimerasa: incorpora nucleótidos complementarios y sintetiza la nueva hebra siempre en dirección 5' → 3'.
- Ligasa: une los fragmentos de ADN, especialmente los fragmentos de Okazaki de la hebra retardada.

##### Apartado 3

Si la ADN polimerasa introduce una base incorrecta y el error no se repara, puede quedar fijado como una mutación en el ADN. Por ejemplo:
- Original:  5' – ATG CCG – 3'
- Mutación:  5' – ATG CTG – 3' (sustitución de una base)

Si esa célula vuelve a replicarse, la modificación puede transmitirse a las células hijas. Sus consecuencias dependerán de dónde ocurra: puede no tener efecto, modificar una proteína o alterar su función.

##### Extensión con Biopython

```python
# Secuencia original escrita en dirección 5' -> 3'
dna = Seq("ATGCCGTTAGCT")
print("Hebra original:")
print(f"5' - {dna} - 3'")


# Hebra complementaria
complementaria = dna.complement()
print("\nHebra complementaria:")
print(f"3' - {complementaria} - 5'")


# Hebra complementaria escrita también en dirección 5' -> 3'
complementaria_5_3 = dna.reverse_complement()
print("\nHebra complementaria escrita 5' -> 3':")
print(f"5' - {complementaria_5_3} - 3'")
```

##### Apartado 2

Para sintetizar ARN, la ARN polimerasa lee la cadena molde en dirección 3' → 5' y fabrica el ARN en dirección 5' → 3'.

Por tanto, en este ejercicio tomamos como cadena molde:
- Cadena codificante:  5' – ATG CCT GAA TGC – 3'
- Cadena molde:        3' – TAC GGA CTT ACG – 5'

La cadena superior se denomina cadena codificante porque su secuencia coincide con la del ARNm, sustituyendo T por U.

Matiz importante: estrictamente, viendo únicamente las dos hebras no podemos saber cuál es la molde. Eso lo determina la posición y orientación del promotor. En este ejercicio se asume que la hebra inferior es la molde.

##### Apartado 3

La ARN polimerasa va leyendo. ADN molde: 3' – TAC GGA CTT ACG – 5' y aplica las complementariedades.

Por tanto:
- ADN molde:   3' – TAC GGA CTT ACG – 5'
- ARNm:        5' – AUG CCU GAA UGC – 3'

El transcrito es:
5' – AUG CCU GAA UGC – 3'

Coincide con la cadena codificante:
- ADN codificante: 5' – ATG CCT GAA TGC – 3'
- ARNm:             5' – AUG CCU GAA UGC – 3'

La única diferencia es: T → U

##### Apartado 4

El promotor es una región de ADN situada antes del gen que permite la unión de la ARN polimerasa y de otros factores necesarios para iniciar la transcripción. En este ejercicio no se proporciona una secuencia promotora, por lo que no podemos señalar exactamente qué nucleótidos constituyen el promotor.

De forma esquemática:
5' ─── PROMOTOR ─── ATG CCT GAA TGC ─── 3'

La secuencia, ATG CCT GAA TGC, puede considerarse el comienzo de una región codificante, ya que comienza por ATG, que corresponde al codón de inicio AUG en el ARNm. Sin embargo, técnicamente no tenemos una región codificante completa porque en la secuencia proporcionada no aparece ningún codón de parada.

##### Extensión con Biopython

```python
# Leer la CDS de HBB desde el archivo FASTA
registro = SeqIO.read("HBB_CDS.fasta", "fasta")
adn = registro.seq
print("ADN codificante:")
print(f"5' - {adn} - 3'")


# Transcribir la cadena codificante de ADN a ARNm
arnm = adn.transcribe()
print("\nARNm:")
print(f"5' - {arnm} - 3'")

print("\nLongitud del ADN:", len(adn), "nucleótidos")
print("Longitud del ARNm:", len(arnm), "nucleótidos")


# --------------------------------------------------
# Experimento cambiando la orientación de la hebra
# --------------------------------------------------

# Obtener la hebra complementaria escrita 5' -> 3'
hebra_molde_5_3 = adn.reverse_complement()
print("\nHebra complementaria/molde escrita 5' -> 3':")
print(f"5' - {hebra_molde_5_3} - 3'")


# Si transcribimos directamente esta hebra, obtenemos otro ARN
arn_orientacion_opuesta = hebra_molde_5_3.transcribe()
print("\nARN obtenido al transcribir directamente la hebra opuesta:")
print(f"5' - {arn_orientacion_opuesta} - 3'")


# Para obtener el ARNm correcto a partir de la hebra molde
# escrita 5' -> 3', debemos hacer primero su reverse complement
arnm_desde_molde = hebra_molde_5_3.reverse_complement().transcribe()
print("\nARNm correcto obtenido a partir de la hebra molde:")
print(f"5' - {arnm_desde_molde} - 3'")


# Comprobar que ambos métodos producen el mismo ARNm
print("\nAmbos ARNm coinciden", arnm == arnm_desde_molde)
```

Biopython obtiene el ARN correspondiente a la cadena codificante sustituyendo las timinas (T) por uracilos (U). Como el archivo utilizado contiene únicamente la CDS de HBB, el resultado representa la región codificante del ARNm y no un ARNm completo con sus regiones UTR. Al cambiar la orientación de la hebra se obtiene una secuencia diferente si se transcribe directamente; sin embargo, al considerar correctamente la orientación de la hebra molde mediante reverse_complement(), se recupera el mismo transcrito.

##### Apartado 2

- AUG es el codón de inicio y codifica para metionina (Met).
- UAA es un codón de parada o STOP. No codifica ningún aminoácido.

##### Apartado 3

Aplicando el código genético:
- AUG → Metionina (Met)
- UAU → Tirosina (Tyr)
- GCU → Alanina (Ala)
- UAA → STOP

Por tanto:
- ARNm: AUG | UAU | GCU | UAA
- Proteína: Met – Tyr – Ala – STOP

La cadena proteica obtenida es: Met – Tyr – Ala.
El codón UAA no forma parte de la proteína; simplemente indica al ribosoma que debe terminar la traducción.

##### Apartado 4

En condiciones normales de traducción eucariota, AUG es el principal codón reconocido como inicio.
Si AUG → GUG, el ribosoma podría no reconocer correctamente el punto de inicio, por lo que la traducción podría no comenzar en ese lugar. Podría buscar otro AUG situado más adelante, dando lugar a una proteína más corta, o incluso no producirse la proteína.

En bacterias, GUG puede actuar como codón de inicio en determinados genes. En estos casos se incorpora normalmente N-formilmetionina (fMet) como aminoácido iniciador.

En caso de que desapareciese el codón STOP, continuaría hasta encontrar otro codón de parada (UAA, UAG o UGA) más adelante. El resultado sería normalmente una proteína más larga de lo normal, que podría tener una estructura y una función alteradas.

##### Apartado 2

El splicing alternativo permite que un mismo gen genere distintos ARNm maduros mediante la combinación diferencial de sus exones. Partiendo de un gen formado por cinco exones se podrían obtener, por ejemplo, las siguientes combinaciones:
- Isoforma 1: Exón 1 – Exón 2 – Exón 4 – Exón 5
- Isoforma 2: Exón 1 – Exón 3 – Exón 5

En ambos casos, determinados exones son excluidos durante el procesamiento del pre-ARNm. De esta forma, a partir de un mismo gen se generan transcritos distintos que pueden dar lugar a proteínas diferentes.

##### Apartado 3

Si esos exones forman parte de la región codificante, las dos moléculas de ARNm tendrán secuencias diferentes y, por tanto, pueden producir proteínas distintas.

Por ejemplo:

- ARNm A: 1 – 2 – 4 – 5 -> Proteína A: [dominio A] [dominio B] [dominio D] [dominio E]
- ARNm B: 1 – 3 – 5     -> Proteína B: [dominio A] [dominio C] [dominio E]

Estas diferencias pueden provocar que las proteínas tengan distinta: longitud, secuencia de aminoácidos, estructura tridimensional, capacidad para unirse a otras moléculas, localización celular y función.

Además, si la exclusión de un exón altera el marco de lectura, podrían cambiar muchos aminoácidos posteriores e incluso aparecer un codón STOP prematuro.

##### Apartado 4

Aumenta la diversidad proteica porque un único gen puede generar varios ARNm maduros, y cada uno puede originar una isoforma proteica diferente. Es decir, no hace falta tener:
- Gen A → proteína A
- Gen B → proteína B
- Gen C → proteína C

Un solo gen puede generar un pre-ARNm que, mediante splicing alternativo, dé lugar a distintos ARNm maduros y, potencialmente, a diferentes isoformas proteicas. Así se aumenta considerablemente la diversidad del proteoma sin aumentar proporcionalmente el número de genes.

##### Extensión con Biopython

```python
ruta = "ENSG00000066468.26.fasta"
transcritos = defaultdict(list)
for registro in SeqIO.parse(ruta, "fasta"):

    # Solo procesamos registros que sean exones de un transcrito
    if "transcript:" not in registro.description or "exon:" not in registro.description:
        continue

    id_transcrito = re.search(
        r"transcript:(\S+)",
        registro.description
    ).group(1)

    numero_exon = int(
        re.search(
            r"exon:(\d+)",
            registro.description
        ).group(1)
    )

    transcritos[id_transcrito].append(
        (numero_exon, registro.seq)
    )


# Reconstruir cada transcrito uniendo sus exones
secuencias = {}
for id_transcrito, exones in transcritos.items():
    exones = sorted(exones, key=lambda x: x[0])
    secuencia = "".join(
        str(secuencia_exon)
        for numero, secuencia_exon in exones
    )

    secuencias[id_transcrito] = secuencia


# Mostrar solo 5 transcritos
print("Número total de transcritos:", len(secuencias))
print("\nPrimeros 5 transcritos de FGFR2:\n")
for id_transcrito in list(secuencias.keys())[:5]:

    print(
        id_transcrito,
        "| Nº exones:", len(transcritos[id_transcrito]),
        "| Longitud:", len(secuencias[id_transcrito]), "nt"
    )


# Comparar dos transcritos
ids = list(secuencias.keys())
t1 = ids[0]
t2 = ids[1]
print("\nComparación de dos transcritos:\n")
print(
    t1,
    "| Nº exones:", len(transcritos[t1]),
    "| Longitud:", len(secuencias[t1]), "nt"
)

print(
    t2,
    "| Nº exones:", len(transcritos[t2]),
    "| Longitud:", len(secuencias[t2]), "nt"
)

if secuencias[t1] == secuencias[t2]:
    print("\nLas secuencias son iguales.")
else:
    print("\nLas secuencias son diferentes.")
```

Los diferentes transcritos de FGFR2 presentan variaciones en la composición y longitud de sus exones, generando distintas isoformas. Estas diferencias pueden modificar la secuencia de aminoácidos y los dominios funcionales de la proteína. En FGFR2, el splicing alternativo puede modificar regiones del dominio extracelular implicadas en la unión a ligandos, alterando la especificidad con la que el receptor reconoce distintos factores de crecimiento. Por tanto, el splicing alternativo permite que un mismo gen genere proteínas con propiedades funcionales diferentes.

##### Apartado 3

El orden de los aminoácidos constituye la estructura primaria de la proteína y determina cómo pueden interaccionar unas regiones con otras. Cada aminoácido tiene propiedades diferentes: algunos son hidrofóbicos, otros hidrofílicos, cargados, polares, etc. Estas propiedades condicionan cómo se pliega la cadena. Por eso, cambiar incluso un solo aminoácido puede modificar el plegamiento y, en algunos casos, la función de la proteína.

No significa que toda mutación hidrofóbico-hidrofílico destruya necesariamente una proteína: el efecto depende mucho de la posición y del papel concreto de ese aminoácido.

##### Apartado 5

El punto más vulnerable a errores que puedan afectar a la función de la proteína es la secuencia de ADN, ya que una mutación puede propagarse al ARNm y, posteriormente, a la proteína. Por ejemplo, una mutación puntual puede cambiar un nucleótido y provocar que un codón pase a codificar un aminoácido diferente. Si ese aminoácido participa en una región importante de la proteína, el cambio puede afectar a su estructura tridimensional, estabilidad o función.

También pueden producirse errores durante la transcripción o la traducción, pero estos suelen afectar únicamente a moléculas concretas de ARN o proteína. En cambio, una mutación permanente en el ADN puede transmitirse a todas las moléculas producidas posteriormente a partir de esa secuencia.

##### Apartado 6 y extensión con Biopython

Se implementa un pipeline completo que integra los procesos anteriores. El programa recibe la CDS de HBB previamente cargada, genera su hebra complementaria, realiza la transcripción, comprueba las características básicas de la región codificante y traduce el ARNm a proteína. Además, informa de cada etapa durante la ejecución.

El pipeline realiza los siguientes procesos:

1. Generar la hebra complementaria del ADN.
2. Obtener el transcrito de ARNm correspondiente.
3. Comprobar que la CDS presenta un codón de inicio, un codón de parada válido y una longitud múltiplo de tres.
4. Traducir el ARNm a una secuencia de aminoácidos.
5. Mostrar un resumen de los resultados obtenidos.

```python
def pipeline_adn_proteina(adn):
    """
    Pipeline completo desde una secuencia codificante de ADN
    hasta la proteína correspondiente.
    """

    print("====================================")
    print("   PIPELINE: DEL ADN A LA PROTEÍNA")
    print("====================================")


    # --------------------------------------------------
    # 1. REPLICACIÓN
    # --------------------------------------------------
    print("\n[1] REPLICACIÓN")

    print("Hebra original:")
    print(f"5' - {adn} - 3'")
    
    complementaria = adn.complement()
    print("Hebra complementaria:")
    print(f"3' - {complementaria} - 5'")


    # --------------------------------------------------
    # 2. TRANSCRIPCIÓN
    # --------------------------------------------------
    print("\n[2] TRANSCRIPCIÓN")

    arnm = adn.transcribe()

    print("ARNm obtenido:")
    print(f"5' - {arnm} - 3'")


    # --------------------------------------------------
    # 3. COMPROBACIÓN DE LA CDS
    # --------------------------------------------------
    print("\n[3] COMPROBACIÓN DE LA CDS")

    codon_inicio = str(adn[:3])
    codon_parada = str(adn[-3:])

    print("Codón de inicio:", codon_inicio)
    print("Codón de parada:", codon_parada)

    if codon_inicio == "ATG":
        print("Presenta el codón de inicio ATG.")
    else:
        print("No presenta ATG como codón inicial.")

    if codon_parada in ["TAA", "TAG", "TGA"]:
        print("Presenta un codón de parada válido.")
    else:
        print("No presenta un codón de parada válido.")

    if len(adn) % 3 == 0:
        print("La longitud de la CDS es múltiplo de 3.")
    else:
        print("La longitud de la CDS no es múltiplo de 3.")


    # --------------------------------------------------
    # 4. TRADUCCIÓN
    # --------------------------------------------------
    print("\n[4] TRADUCCIÓN")

    proteina = arnm.translate(cds=True)

    print("Proteína obtenida:")
    print(proteina)


    # --------------------------------------------------
    # 5. RESUMEN
    # --------------------------------------------------
    print("\n======= PIPELINE COMPLETADO =======")

    print(f"Longitud del ADN:      {len(adn)} nucleótidos")
    print(f"Longitud del ARNm:     {len(arnm)} nucleótidos")
    print(f"Longitud de proteína:  {len(proteina)} aminoácidos")

    print("ADN → replicación → ARNm → proteína")

    return arnm, proteina


arnm_final, proteina_final = pipeline_adn_proteina(adn)
```
